```python
#Task 3.1

def inputPuzzle():
    count = 0
    listrows = []
    while count != 9:
        valid = False
        row = input("Enter row: ").split(' ')
        for number in row:
            if number in '123456789' and len(row) == 9 and len(number) == 1:
                valid = True

            else:
                valid = False
                break
        if valid == True:
            print("Row successfully added")
            count += 1
            listrows.append(row)
        else:
            print("Please input a valid row")
            
    if count == 9:        
        return listrows

puzzle1 = inputPuzzle() #Try Example1

puzzle2 = inputPuzzle() #Try Example2
```

```python
#TASK 3.2
def horizontal_check(puzzle):
    
    count = 0
    for row in puzzle:
        count += 1
        rowcheck = ''
        for number in row:
            if number in rowcheck:
                return f"Duplicate in row {count}"
                break
            rowcheck += number
    return -1

#Display Check
def display_row(puzzle):
    if puzzle == -1:
        return "Row Valid!"
    else:
        return puzzle

#TEST CASE
print(display_row(horizontal_check(puzzle1))) #puzzle1
print(display_row(horizontal_check(puzzle2))) #puzzle2
```

```python
#Task 3.3
def column_check(puzzle):
    count = 0
    columncheck = ''
    while count != 9:
        for row in puzzle:
            if row[count] in columncheck:
                return f"Duplicate in column {count+1} (column: {columncheck + row[count]})"
            else:
                columncheck += row[count]

            if len(columncheck) == 9:
                columncheck = ''
                count += 1
    return -1
    
#Test Case
def display_column(puzzle):
    if puzzle == -1:
        return "Column Valid!"
    else:
        return puzzle

#TEST CASE
print(display_column(column_check(puzzle1))) #puzzle1
print(display_column(column_check(puzzle2))) #puzzle2
```

```python
#Task 3.4

def mini_square_check(puzzle):
    count_row = 0
    count_column = 0
    count = 0
    square_check = ''
    while count_row < 9:
        for i in range(3):
            for j in range(3):
                
                if puzzle[i+count_row][j+count_column] not in square_check:
                    square_check += puzzle[i+count_row][j+count_column]
                else:
                    return False, count + 1
                    
            
            
        if len(square_check) == 9:
            square_check = ''
            count_row += 3
            count += 1
            
            if i + count_row > 8:
                count_row = 0
                count_column += 3
            if i + count_row > 8 or j + count_column > 8:
                return True

def display_minisqcheck(puzzle):
    if mini_square_check(puzzle) is True:
        return "Sudoku!"
    return f"Invalid Sudoku at {mini_square_check(puzzle)[1]}"

#TEST CASE
print(display_minisqcheck(puzzle1))
print(display_minisqcheck(puzzle2))

```
