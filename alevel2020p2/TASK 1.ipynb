```python
#Task 1.1
def task1_1(string_value):
    h = 0
    n = len(string_value)
    for i in range(n-1):
        val = 33 * ord(string_value[i])
        h = (h + val) % 1024
    return h
print(task1_1("Hello")) #3 call test
print(task1_1("Hallo"))
print(task1_1("Hullo"))
```

```python
#Task 1.2
def task1_2(seed,string_value):
    ss = seed + string_value
    return task1_1(ss)
    
print(task1_2("seed—one", "Hello")) # 3call test
print(task1_2("seed—two", "Hello"))
print(task1_2("seed—three", "Hello"))
```
