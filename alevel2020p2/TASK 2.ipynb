```python
#Task 2.1
import random
def task2_1(filename,quantity,maximum):
    with open(filename,'w') as file: #file auto closes
        for i in range(quantity):
            i = random.randint(0,maximum)
            file.write(f"{i}\n")
            
task2_1("randomnumbers.txt",1000,5000)         
```

```python
#Task 2.2

def task2_2(list_of_integers):
    def mergesort(list_of_integers):
        n = len(list_of_integers)
        arr = list_of_integers
        if n > 1:
            left_arr = arr[:n//2]
            right_arr = arr[n//2:]
            mergesort(left_arr)
            mergesort(right_arr)
            i = 0
            j = 0
            k = 0
            while i < len(left_arr) and j < len(right_arr):
                if left_arr[i] < right_arr[j]:
                    arr[k] = left_arr[i]
                    i += 1
                else:
                    arr[k] = right_arr[j]
                    j += 1
                k += 1
            while i < len(left_arr):
                arr[k] = left_arr[i]
                i += 1
                k += 1
            while j < len(right_arr):
                arr[k] = right_arr[j]
                j += 1
                k += 1
        return arr
    return mergesort(list_of_integers)

# test function
task2_2([56,25,4,98,0,18,4,5,7,0]) == [0,0,4,4,5,7,18,25,56,98]
```

```python
#Task 2.3
def task2_3(filename_in, filename_out):
    with open(filename_in,'r') as file: #automatic close
        lst = []
        for line in file:
            line = file.readline().strip()
            lst.append(int(line))
            task2_2(lst)
    with open(filename_out,'w') as file:
        for i in lst:
            file.write(f"{i}\n")

task2_3("randomnumbers.txt","sortednumbers.txt")
            
```
