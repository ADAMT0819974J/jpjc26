```python
#TASK 2.1
class Node:
    def __init__(self):
        self.data_value = -1
        self.left_node = -1
        self.right_node = -1
        
    def set_data(self,data):
        self.data_value = data

    def set_left(self,value):
        self.left_node = value

    def set_right(self,value):
        self.right_node = value

    def get_data(self):
        return self.data_value

    def get_left(self):
        return self.left_node

    def get_right(self):
        return self.right_node  

        
```

```python
#Task 2.2
def calculate_nodes(numlevel): 
    if numlevel == 1:
        return numlevel
    return 2**numlevel - 1
```

```python
#Task 2.3

print(f"""
Test 1: {calculate_nodes(1)}

Test 2: {calculate_nodes(3)}

Test 3: {calculate_nodes(5)}

Test 4: {calculate_nodes(10)}
""")
```

```python
#Task 2.4
def create_empty_tree(listnodes,numlevel):
    nlist = [' '] * calculate_nodes(numlevel)
    for i in range(len(listnodes)):
        nlist[i] = Node()
    return nlist

emptree = create_empty_tree(['','','','','','',''],3)
print(emptree)
```

```python
#Task 2.5
def assign_leaf_nodes(listnodes,numlevel):
    for i in range(2**(numlevel-1)):
        listnodes[i].set_data(i+1)

assign_leaf_nodes(emptree,3)
```

```python
#Task 2.6
def create_number_tree(listnodes,numlevel):
    c = 0 #use c to traverse
    for i in range(len(listnodes)):
        if listnodes[i].get_data() == -1:
            listnodes[i].set_data(listnodes[c].get_data() + listnodes[c+1].get_data()) 
            listnodes[i].set_left(c)
            listnodes[i].set_right(c+1)
            c += 2
    return listnodes

numtree = create_number_tree(emptree,3)
for i in range(len(numtree)):
    print(numtree[i].get_data())
```

```python
#Task 2.7
def display(level):
    print(f"""
    Level {level}
    No. Nodes: {calculate_nodes(1)}
    """)
    emtree = create_empty_tree(['']*calculate_nodes(level),level)
    data= []
    for i in emtree:
        data.append(i.get_data())
    print(f"""
    Empty Tree: {data}
    """)
    assign_leaf_nodes(emtree,level)
    data = []
    for i in emtree:
        data.append(i.get_data())
    print(f"""
    Tree with Leaf Nodes Only: {data}
    """)

    numtree = create_number_tree(emtree,level)
    print(f"""
    Full Tree:""")
    for i in numtree:
        print(f"""
        {i.get_left()}\t{i.get_data()}\t{i.get_right()}
        """)
#Test Cases
display(1)
display(3)
display(5)
```
