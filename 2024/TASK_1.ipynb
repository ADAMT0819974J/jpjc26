```python
#TASK 1.1
import random
def hand():
    hand = []
    colour = ['r','g','b']
    for i in range(10):
        hand.append(colour[random.randint(0,2)] + str(random.randint(1,4)))
    return hand

p1 = hand()
p2 = hand()
print(f"""
Player 1: {p1}

Player 2: {p2}
""")
```

```python
def compare(c1,c2):
    if c1 == c1:
        return 3 #same
    elif c1 == 'r':
        if c2 == 'g': #c2 lose
            return 1
        elif c2 == 'b': #c2 win
            return 2
    elif c1 == 'b':
        if c2 == 'r':
            return 1 #c1 win
        elif c2 == 'g':
            return 2 #c1 lose
    elif c1 == 'g':
        if c2 == 'r':
            return 2 #c2 win
        elif c2 == 'b':
            return 1 #c1 win        
```

```python
def handsort(hand):
    n = len(hand)
    order = {
        'r':1,
        'g':2,
        'b':3}
    if n > 1:
        left_hand = hand[:n//2]
        right_hand = hand[n//2:]

        handsort(left_hand)
        handsort(right_hand)

        i = 0
        j = 0
        k = 0

        while i < len(left_hand) and j < len(right_hand):
            
            if left_hand[i][1] < right_hand[j][1]:
                hand[k] = left_hand[i]
                i += 1
            elif left_hand[i][1] > right_hand[j][1]:
                hand[k] = right_hand[j]
                j += 1
            elif order[left_hand[i][0]] < order[right_hand[j][0]]:
                hand[k] = left_hand[i]
                i += 1
            else:
                hand[k] = right_hand[j]
                j += 1
            k += 1

        while i < len(left_hand):
            hand[k] = left_hand[i]
            i += 1
            k += 1

        while j < len(right_hand):
            hand[k] = right_hand[j]
            j += 1
            k += 1
        return hand

print(f"""
Before: {p1}\n
After: {handsort(p1)}\n

Before: {p2}\n
After: {handsort(p2)}\n
""")
```

```python
#TASK 1.4

def calculate(p1,p2):
    p1 = mergesort(p1)
    p2 = mergesort(p2)
    scorep1 = 0
    scorep2 = 0
    for i in range(10):
        if compare(p1[i][0],p2[i][0]) == 1:
            scorep1 += int(p1[i][1] * 2)
            scorep2 += int(p2[i][1])
        elif compare(p1[i][0],p2[i][0]) == 2:
            scorep2 += int(p2[i][1] * 2)
            scorep1 += int(p1[i][1])
        else:
            scorep1 += int(p1[i][1])
            scorep2 += int(p2[i][1])
    return scorep1,scorep2

print(f"""
Sorted Hand: {mergesort(p1)}
Score Player 1: {calculate(p1,p2)[0]}

Sorted Hand: {mergesort(p2)}
Score Player 2: {calculate(p1,p2)[1]}
""")
            
```

```python
def checkwin(p1,p2):
    if calculate(p1,p2)[0] > calculate(p1,p2)[1]:
        print(f"""
        Player 1 Win! Score {calculate(p1,p2)[0]}
        Player 2 Lost! Score {calculate(p1,p2)[1]}""")
        return
        
    elif calculate(p1,p2)[0] < calculate(p1,p2)[1]:
        print(f"""
        Player 2 Win! Score {calculate(p1,p2)[1]}
        Player 1 Lost! Score {calculate(p1,p2)[0]}""")
        return
    print(f"""
    Draw! Both win
    Score:{calculate(p1,p2)[0]}""")
    return

checkwin(p1,p2)
    
```
