```python
#Task 3.1
def format_heading(text):
    ftext = f"<h1>{text}</h1>"
    return ftext

def format_subheading(text):
    ftext = f"<h2>{text}</h2>"
    return ftext

def format_para(text):
    ftext = f"<p1>{text}</p1>"
    return ftext
#Test Cases
text = "My first page"
print(f"""
heading -> {format_heading(text)}

subheading -> {format_subheading(text)}

paragraph -> {format_para(text)}
""")
    
```

```python
def hyperlink(url,string):
    ftext = f"<a href={url}>{string}<a>"
    return ftext

print(f"""
hyperlink -> {hyperlink("https://www. cambridgeinternational. org","Click here")}
""")
    
```

```python
#Task 3.3
def body(textlist):
    fstring = ''
    for i in range(len(textlist)):
        part = textlist[i].split(': ')
        if '\n' in part[1]:
            part[1] = part[1][:len(part[1])-1]
        if part[0].lower() == 'heading':
            fstring += format_heading(part[1]) +'\n' +'\n'
        elif part[0].lower() == 'subheading':
            fstring += format_subheading(part[1]) + '\n'+'\n'
        elif part[0].lower() == 'paragraph':
            fstring += format_para(part[1]) + '\n'+'\n'
        elif part[0].lower() == 'hyperlink':
            h = part[1].split(',')
            fstring += hyperlink(h[0],h[1])
        else:
            return False
    return f"<html>\n\n{fstring}</html>"
print(body(['heading: Myname\n']))
```

```python
#Task 3.4
def read_data(filename):
    with open(filename,'r') as file: #auto closes
        lines = file.readlines()
        fline = body(lines)
    return fline

def write_data(filename,fstring):
    if fstring == 'False':
        return "Error: Can't be converted to html"
    with open(f"{filename}.html","w") as file: #autocloses
        file.writelines(f"""
        <DOCTYPE! html>
        {fstring}
        """)


#Test case

write_data("test",read_data("firstPage.txt"))
#Test case 2

write_data("theWebsite",read_data("website.txt"))
```
